```python
import requests
import json
```

```python
api_key = "" # fill in your API key here
```

```python
uri = "http://140.120.13.248:8787/ollama/"
headers = {
    "X-API-Key": api_key
}
chat_api_url = f"{uri}/api/chat"
```

```python
def generate(model: str, messages: list, stream: bool, options: dict):
  payload = json.dumps({
    "model": model,
    "messages": messages,
    "stream": stream,
    "options": options
  })
  headers = {
    "X-API-Key": api_key,
    'Content-Type': 'application/json',
    'accept': 'application/json'
  }

  response = requests.request("POST", chat_api_url, headers=headers, data=payload)

  return response.json()
```

```python
model = "gpt-oss:20b"
stream = False
messages = [
  {
    "role": "user",
    "content": "你是誰"
  }
]
options = {
  "temperature": 1.0,
  # "max_new_tokens": 10
}
```

```python
res = generate(model, messages, stream, options)
```

```python
print(f"===== Reason =====\n{res['message']['thinking']}")
print(f"===== Answer =====\n{res['message']['content']}")
```
